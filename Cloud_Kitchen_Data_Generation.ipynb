```python
# ============================================
# STEP 4.8.5: Calculate Restaurant Selection Weights
# ============================================

np.random.seed(42)

# Start with the restaurant's estimated demand
restaurant_weights = restaurants["estimated_monthly_orders"].astype(float).copy()

# Rating effect
restaurant_weights *= (
    restaurants["rating"] / restaurants["rating"].mean()
)

# Discount effect
restaurant_weights *= (
    1 + restaurants["discount_pct"] / 200
)

# Faster delivery gets a small advantage
delivery_factor = (
    restaurants["avg_delivery_time_min"].mean()
    / restaurants["avg_delivery_time_min"]
)

restaurant_weights *= delivery_factor

# Make sure all weights are positive
restaurant_weights = restaurant_weights.clip(lower=0.01)

# Convert to probabilities
restaurant_probabilities = (
    restaurant_weights / restaurant_weights.sum()
)

print("Total restaurant probability:",
      restaurant_probabilities.sum())

print("\nHighest-weight restaurants:")

print(
    restaurants[
        ["restaurant_id", "cuisine", "rating",
         "estimated_monthly_orders"]
    ]
    .assign(selection_probability=restaurant_probabilities)
    .sort_values(
        "selection_probability",
        ascending=False
    )
    .head(10)
)
```

```python
# ============================================
# STEP 4.8.6: Create Customer-Restaurant Matching
# ============================================

# Create a lookup table for restaurant information
restaurant_lookup = restaurants[
    [
        "restaurant_id",
        "locality_id",
        "cuisine",
        "rating",
        "estimated_monthly_orders",
        "avg_delivery_time_min",
        "discount_pct"
    ]
].copy()

# Base restaurant score
restaurant_lookup["base_score"] = (
    restaurant_lookup["estimated_monthly_orders"]
    * (restaurant_lookup["rating"] / restaurants["rating"].mean())
)

# Create a function to calculate restaurant probabilities
def get_restaurant_probabilities(customer_locality, preferred_cuisine):

    scores = restaurant_lookup["base_score"].copy()

    # Same locality gets a strong boost
    scores = scores * np.where(
        restaurant_lookup["locality_id"] == customer_locality,
        3.0,
        0.5
    )

    # Preferred cuisine gets a strong boost
    scores = scores * np.where(
        restaurant_lookup["cuisine"] == preferred_cuisine,
        3.0,
        0.7
    )

    # Faster delivery gets a small advantage
    delivery_factor = (
        restaurant_lookup["avg_delivery_time_min"].mean()
        / restaurant_lookup["avg_delivery_time_min"]
    )

    scores = scores * delivery_factor

    # Discount gets a small advantage
    scores = scores * (
        1 + restaurant_lookup["discount_pct"] / 200
    )

    # Convert scores into probabilities
    probabilities = scores / scores.sum()

    return probabilities.values


# Test the matching for the first customer
test_customer = customers.iloc[0]

test_probabilities = get_restaurant_probabilities(
    test_customer["locality_id"],
    test_customer["preferred_cuisine"]
)

# Show top 5 restaurant choices for this customer
test_results = restaurant_lookup[
    ["restaurant_id", "locality_id", "cuisine"]
].copy()

test_results["probability"] = test_probabilities

test_results = test_results.sort_values(
    "probability",
    ascending=False
).head(5)

print("Test customer:")
print(test_customer[
    ["customer_id", "locality_id",
     "customer_segment", "preferred_cuisine"]
])

print("\nMost likely restaurants:")
print(test_results)
```

```python
# ============================================
# STEP 4.8.7: Assign Restaurants to Orders
# ============================================

np.random.seed(42)

# Randomly select customers for each order
selected_customer_indices = np.random.choice(
    customers.index,
    size=n_orders,
    p=customers["order_probability"]
)

# Get the selected customer records
selected_customers = customers.loc[
    selected_customer_indices
].reset_index(drop=True)

# Store customer IDs
order_customer_ids = selected_customers["customer_id"].values

# Store customer localities
order_localities = selected_customers["locality_id"].values

# Store customer segments
order_segments = selected_customers["customer_segment"].values

# Store preferred cuisines
order_preferred_cuisines = selected_customers["preferred_cuisine"].values


# --------------------------------------------
# Function to select a restaurant
# --------------------------------------------

def select_restaurant(locality, preferred_cuisine):

    # Start with base restaurant score
    scores = restaurant_lookup["base_score"].copy()

    # Same locality
    scores *= np.where(
        restaurant_lookup["locality_id"] == locality,
        3.0,
        0.5
    )

    # Preferred cuisine
    scores *= np.where(
        restaurant_lookup["cuisine"] == preferred_cuisine,
        3.0,
        0.7
    )

    # Delivery speed
    scores *= (
        restaurant_lookup["avg_delivery_time_min"].mean()
        / restaurant_lookup["avg_delivery_time_min"]
    )

    # Discount
    scores *= (
        1 + restaurant_lookup["discount_pct"] / 200
    )

    # Convert to probabilities
    probabilities = scores / scores.sum()

    # Select restaurant
    return np.random.choice(
        restaurant_lookup["restaurant_id"],
        p=probabilities
    )


# Select restaurants for all orders
order_restaurant_ids = [
    select_restaurant(locality, cuisine)
    for locality, cuisine
    in zip(
        order_localities,
        order_preferred_cuisines
    )
]


# Check results
print("Restaurants assigned:", len(order_restaurant_ids))

print("\nFirst 10 restaurant assignments:")
print(order_restaurant_ids[:10])
```

```python
# ============================================
# STEP 4.8.8: Select Menu Items for Orders
# ============================================

np.random.seed(42)

# Create a quick lookup of menu items by restaurant
restaurant_menu = {}

for restaurant_id, group in menu_items.groupby("restaurant_id"):

    restaurant_menu[restaurant_id] = {
        "item_ids": group["item_id"].values,
        "popularity": group["popularity_score"].values
    }


# --------------------------------------------
# Function to select an item from a restaurant
# --------------------------------------------

def select_menu_item(restaurant_id):

    menu = restaurant_menu[restaurant_id]

    # Use popularity score as selection weight
    weights = menu["popularity"].astype(float)

    # Convert to probabilities
    probabilities = weights / weights.sum()

    # Select an item
    return np.random.choice(
        menu["item_ids"],
        p=probabilities
    )


# --------------------------------------------
# Select item for every order
# --------------------------------------------

order_item_ids = [
    select_menu_item(restaurant_id)
    for restaurant_id in order_restaurant_ids
]


# --------------------------------------------
# Check results
# --------------------------------------------

print("Menu items assigned:", len(order_item_ids))

print("\nFirst 10 item assignments:")
print(order_item_ids[:10])
```

```python
# ============================================
# STEP 4.8.10B: Calculate Order Value
# ============================================

np.random.seed(42)

# --------------------------------------------
# Get discount percentage for each restaurant
# --------------------------------------------

restaurant_discount_lookup = restaurants[
    ["restaurant_id", "discount_pct"]
].copy()

selected_items = selected_items.merge(
    restaurant_discount_lookup,
    on="restaurant_id",
    how="left"
)

# Check for missing discounts
print("Missing restaurant discounts:")
print(selected_items["discount_pct"].isnull().sum())


# --------------------------------------------
# Calculate discount amount
# --------------------------------------------

selected_items["discount_amount"] = (
    selected_items["price"]
    * selected_items["discount_pct"]
    / 100
)


# --------------------------------------------
# Calculate discounted food amount
# --------------------------------------------

selected_items["food_amount"] = (
    selected_items["price"]
    - selected_items["discount_amount"]
)


# --------------------------------------------
# Generate delivery/platform fee
# --------------------------------------------

delivery_fees = np.random.choice(
    [20, 25, 30, 35, 40],
    size=n_orders,
    p=[0.15, 0.25, 0.30, 0.20, 0.10]
)

selected_items["delivery_fee"] = delivery_fees


# --------------------------------------------
# Final amount paid by customer
# --------------------------------------------

selected_items["order_value"] = (
    selected_items["food_amount"]
    + selected_items["delivery_fee"]
)


# --------------------------------------------
# Check results
# --------------------------------------------

print("\nOrder value summary:")
print(
    selected_items[
        [
            "price",
            "discount_pct",
            "discount_amount",
            "food_amount",
            "delivery_fee",
            "order_value"
        ]
    ].describe()
)

print("\nSample:")
print(
    selected_items[
        [
            "item_id",
            "item_name",
            "price",
            "discount_pct",
            "discount_amount",
            "delivery_fee",
            "order_value"
        ]
    ].head(10)
)
```

```python
# ============================================
# STEP 4.8.11: Assign Order Status
# ============================================

# Use a separate random generator
rng_status = np.random.default_rng(2025)

order_status = []

for platform in order_platforms:

    if platform == "Swiggy":

        status = rng_status.choice(
            ["Delivered", "Cancelled", "Failed"],
            p=[0.94, 0.04, 0.02]
        )

    elif platform == "Zomato":

        status = rng_status.choice(
            ["Delivered", "Cancelled", "Failed"],
            p=[0.94, 0.04, 0.02]
        )

    else:  # Direct

        status = rng_status.choice(
            ["Delivered", "Cancelled", "Failed"],
            p=[0.70, 0.20, 0.10]
        )

    order_status.append(status)


order_status = np.array(order_status)


# --------------------------------------------
# Check results
# --------------------------------------------

print("Order statuses assigned:", len(order_status))

print("\nOrder status distribution:")
print(
    pd.Series(order_status)
    .value_counts()
)

print("\nOrder status percentage:")
print(
    pd.Series(order_status)
    .value_counts(normalize=True)
    .mul(100)
    .round(2)
)
```

```python
# ============================================
# STEP 4.8.12: Assign Customer Ratings
# ============================================

np.random.seed(42)

# Start with all ratings as missing
order_ratings = np.full(
    n_orders,
    np.nan
)

# Get restaurant ratings for each order
order_restaurant_ratings = restaurants.set_index(
    "restaurant_id"
).loc[
    order_restaurant_ids,
    "rating"
].values

# Identify delivered orders
delivered_mask = (
    np.array(order_status) == "Delivered"
)

# Generate ratings only for delivered orders
base_ratings = (
    order_restaurant_ratings[delivered_mask]
    + np.random.normal(
        loc=0,
        scale=0.6,
        size=delivered_mask.sum()
    )
)

# Round to nearest whole number
generated_ratings = np.round(
    base_ratings
)

# Keep ratings between 1 and 5
generated_ratings = np.clip(
    generated_ratings,
    1,
    5
)

# Assign ratings
order_ratings[delivered_mask] = generated_ratings

# Check results
print("Ratings generated:", delivered_mask.sum())

print("\nMissing ratings:")
print(
    np.isnan(order_ratings).sum()
)

print("\nRating distribution:")
print(
    pd.Series(order_ratings)
    .value_counts(dropna=False)
    .sort_index()
)
```

```python
# ============================================
# STEP 4.36: CUSTOMER BEHAVIOR SEGMENTATION
# ============================================

def classify_customer(row):

    orders_count = row["total_orders"]
    revenue = row["total_revenue"]
    success_rate = row["delivery_success_rate"]

    # High-value loyal customers
    if orders_count >= 10 and revenue >= 1800:
        return "High Value Loyal"

    # Frequent customers
    elif orders_count >= 8:
        return "Frequent Customer"

    # High spenders with fewer orders
    elif revenue >= 1500:
        return "High Value Occasional"

    # Regular customers with good delivery success
    elif orders_count >= 4 and success_rate >= 90:
        return "Regular Customer"

    # Low activity customers
    else:
        return "Low Activity"


# --------------------------------------------
# Assign behavioral segment
# --------------------------------------------

customer_behavior["behavior_segment"] = (
    customer_behavior.apply(
        classify_customer,
        axis=1
    )
)


# --------------------------------------------
# Display customer behavior segments
# --------------------------------------------

print("CUSTOMER BEHAVIOR SEGMENTATION")
print("=" * 100)

print(
    customer_behavior[
        [
            "total_orders",
            "total_revenue",
            "average_order_value",
            "delivery_success_rate",
            "behavior_segment"
        ]
    ].head(20)
)


# --------------------------------------------
# Segment distribution
# --------------------------------------------

print("\n\nBEHAVIOR SEGMENT DISTRIBUTION")
print("=" * 100)

print(
    customer_behavior["behavior_segment"]
    .value_counts()
)


# --------------------------------------------
# Segment performance
# --------------------------------------------

behavior_segment_summary = (
    customer_behavior
    .groupby("behavior_segment")
    .agg(
        customers=("total_orders", "count"),

        total_orders=("total_orders", "sum"),

        total_revenue=("total_revenue", "sum"),

        average_orders=("total_orders", "mean"),

        average_revenue=("total_revenue", "mean"),

        average_order_value=("average_order_value", "mean"),

        average_rating=("average_rating", "mean"),

        average_delivery_success=(
            "delivery_success_rate",
            "mean"
        )
    )
    .sort_values(
        "total_revenue",
        ascending=False
    )
    .round(2)
)


# --------------------------------------------
# Display segment performance
# --------------------------------------------

print("\n\nBEHAVIOR SEGMENT PERFORMANCE")
print("=" * 100)

print(behavior_segment_summary)
```

```python
# ============================================
# STEP 4.37 — CUSTOMER PURCHASE FREQUENCY & RECENCY
# ============================================

# Make sure order_date is datetime
orders["order_date"] = pd.to_datetime(orders["order_date"])

# Latest date in the dataset
latest_order_date = orders["order_date"].max()

print("Latest order date:", latest_order_date)

# --------------------------------------------
# Customer purchase behavior
# --------------------------------------------

customer_frequency = (
    orders
    .groupby("customer_id")
    .agg(
        first_order_date=("order_date", "min"),
        last_order_date=("order_date", "max"),
        total_orders=("order_id", "nunique"),
        total_revenue=("order_value", "sum")
    )
)

# --------------------------------------------
# Recency
# --------------------------------------------

customer_frequency["recency_days"] = (
    latest_order_date - customer_frequency["last_order_date"]
).dt.days

# --------------------------------------------
# Customer lifetime
# --------------------------------------------

customer_frequency["customer_lifetime_days"] = (
    customer_frequency["last_order_date"]
    - customer_frequency["first_order_date"]
).dt.days

# --------------------------------------------
# Average days between orders
# --------------------------------------------

customer_frequency["average_days_between_orders"] = (
    customer_frequency["customer_lifetime_days"]
    / (customer_frequency["total_orders"] - 1)
).where(
    customer_frequency["total_orders"] > 1,
    0
)

# --------------------------------------------
# Activity status
# --------------------------------------------

customer_frequency["activity_status"] = pd.cut(
    customer_frequency["recency_days"],
    bins=[-1, 30, 90, 180, float("inf")],
    labels=[
        "Active",
        "Recently Inactive",
        "At Risk",
        "Inactive"
    ]
)

# --------------------------------------------
# Round values
# --------------------------------------------

customer_frequency = customer_frequency.round({
    "total_revenue": 2,
    "average_days_between_orders": 2
})

# --------------------------------------------
# Display
# --------------------------------------------

print("CUSTOMER PURCHASE FREQUENCY & RECENCY")
print("=" * 100)

print(customer_frequency.head(20))
```

## Locality Opportunity Score Methodology

The locality opportunity score was developed to compare potential cloud-kitchen locations using seven factors related to market potential, customer fit, and competition. Each factor was assigned a weight based on its expected importance for evaluating a new cloud-kitchen location.

| Factor | Weight | Rationale |
|---|---:|---|
| Population Density | 20% | Higher population density indicates a larger potential customer base within the delivery area. |
| Average Monthly Income | 20% | Higher income levels can support greater food-delivery spending and higher-value orders. |
| Student Share | 10% | Students represent an important target segment for affordable and convenient food delivery. |
| Office Worker Share | 15% | Office workers can generate consistent weekday lunch and dinner demand. |
| Residential Share | 10% | Higher residential concentration supports recurring household delivery demand. |
| Tourism Score | 10% | Tourism can provide additional demand, although it may be less consistent than residential or working-population demand. |
| Low Competition | 15% | Lower competition provides a better opportunity for a new cloud kitchen to enter and establish market presence. |
| **Total** | **100%** | |

```python
# ============================================
# STEP 8.1: LOCALITY OPPORTUNITY SCORING
# ============================================

locality_opportunity = localities.copy()

# Normalize positive opportunity factors to 0–100
def min_max_score(series):
    return (
        (series - series.min()) /
        (series.max() - series.min()) * 100
    )

locality_opportunity["density_score"] = min_max_score(
    locality_opportunity["population_density"]
)

locality_opportunity["income_score"] = min_max_score(
    locality_opportunity["avg_monthly_income"]
)

locality_opportunity["student_score"] = min_max_score(
    locality_opportunity["student_share_pct"]
)

locality_opportunity["office_score"] = min_max_score(
    locality_opportunity["office_worker_share_pct"]
)

locality_opportunity["residential_score"] = min_max_score(
    locality_opportunity["residential_share_pct"]
)

locality_opportunity["tourism_score_normalized"] = min_max_score(
    locality_opportunity["tourism_score"]
)

# Lower competition is better for a new entrant
locality_opportunity["competition_opportunity_score"] = (
    100 - min_max_score(locality_opportunity["competition_score"])
)

# --------------------------------------------
# Weighted opportunity score
# --------------------------------------------

locality_opportunity["opportunity_score"] = (
    locality_opportunity["density_score"] * 0.20 +
    locality_opportunity["income_score"] * 0.20 +
    locality_opportunity["student_score"] * 0.10 +
    locality_opportunity["office_score"] * 0.15 +
    locality_opportunity["residential_score"] * 0.10 +
    locality_opportunity["tourism_score_normalized"] * 0.10 +
    locality_opportunity["competition_opportunity_score"] * 0.15
)

# Rank localities
locality_opportunity["opportunity_rank"] = (
    locality_opportunity["opportunity_score"]
    .rank(ascending=False, method="min")
    .astype(int)
)

# Round values
locality_opportunity = locality_opportunity.round(2)

# Sort by opportunity
locality_opportunity = locality_opportunity.sort_values(
    "opportunity_score",
    ascending=False
)

print("LOCALITY OPPORTUNITY SCORING")
print("=" * 100)

print(
    locality_opportunity[
        [
            "locality_name",
            "zone",
            "population_density",
            "avg_monthly_income",
            "student_share_pct",
            "office_worker_share_pct",
            "tourism_score",
            "competition_score",
            "opportunity_score",
            "opportunity_rank"
        ]
    ]
)
```
